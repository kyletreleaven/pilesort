```python
import random

N = 50
seq = list(range(N))
random.shuffle(seq)

seq
```

```python
from setiptah.pilesort.util import singleton, invert_perm

THREE_QUEUES = "QQQ"

@singleton
class THREE_QUEUES_ARBITRARILY_MANY_TIMES:
    def __iter__(self):
        while True:
            yield THREE_QUEUES

position_form = invert_perm(seq)

shuffle = minsort_multiround(position_form, THREE_QUEUES_ARBITRARILY_MANY_TIMES)

shuffle
```

```python
from setiptah.pilesort.shuffle import deal_values, collect_piles

print("start hand", seq); print()
result = seq
for round_idx, shuffle_ in enumerate(shuffle):
    print(f"round {round_idx + 1}")
    piles = deal_values(result, shuffle_, len(THREE_QUEUES))
    print("piles", piles)

    result = collect_piles(piles, THREE_QUEUES)
    print("hand", result); print()
```

```python
from setiptah.pilesort.util import singleton, invert_perm

@singleton
class random_horizon:

    def __init__(self):
        self.result = []
    
    def __iter__(self):
        yield from self.result
        while True:
            round = "".join(random.choice("QS") for _ in range(3))
            self.result.append(round)
            yield round

position_form = invert_perm(seq)

shuffle = minsort_multiround(position_form, random_horizon)

shuffle, random_horizon.result
```

```python
from setiptah.pilesort.shuffle import deal_values, collect_piles

print("start hand", seq); print()
result = seq
for round_idx, (shuffle_, pile_types) in enumerate(zip(shuffle, random_horizon.result)):
    print(f"round {round_idx + 1}")
    piles = deal_values(result, shuffle_, len(pile_types))
    print("piles", piles)

    result = collect_piles(piles, pile_types)
    print("hand", result); print()
```
