# Sorting Optimally with Piles

This library provides simple facilities for sorting optimally with piles.

```python
import random

seq = list(range(20))
random.shuffle(seq)

seq
```

```python
from setiptah.pilesort.minsort import minsort

minsort?
```

```python
from setiptah.pilesort.shuffle import QUEUES
from setiptah.pilesort.util import invert_perm

position_form = invert_perm(seq)
shuffle, pile_types = minsort(position_form, QUEUES)

pile_types, shuffle
```

```python
from setiptah.pilesort.shuffle import deal_values

deal_values?
```

```python
piles = deal_values(seq, shuffle, len(pile_types))

piles
```

```python
from setiptah.pilesort.shuffle import collect_piles

collect_piles?
```

```python
shuffled = collect_piles(piles, pile_types)

shuffled
```

```python
from setiptah.pilesort.shuffle import shuffle_values

shuffle_values?
```

```python
shuffle_values(seq, shuffle, pile_types)
```
